```python
from transformer_lens import HookedTransformer
import torch
```

```python
model = HookedTransformer.from_pretrained("gemma-2b")
```

```python
text = "Donald Trump is"

tokens = model.to_tokens(text) # encode
```

```python
logits, activations = model.run_with_cache(tokens) # run with activations saved
```

```python
# 假设 logits 是一个 torch.Tensor，形状为 (batch_size, sequence_length, vocab_size)

# 1. 提取最后一个时间步的 logits（形状：(batch_size, vocab_size)）
last_step_logits = logits[:, -1, :]

# 2. 对最后一个时间步的 logits 沿着词汇表维度 (dim=-1) 求 argmax
#    这将得到预测的 token IDs (形状：(batch_size,))
predicted_tokens = torch.argmax(last_step_logits, dim=-1)

# 3. 将 Token IDs 传递给 to_string 进行解码
print(model.to_string(predicted_tokens))
```

```python
complete_response_tokens = model.generate(tokens)

model.to_string(complete_response_tokens)
```
